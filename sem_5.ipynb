# Статистика в словарях

````python
purchases = [("Еда", 300), ("Транспорт", 100), ("Еда", 200)]
expenses = {}

for category, amount in purchases:
    if category not in expenses:
        expenses[category] = 0
    expenses[category] += amount

print(expenses) # Что выведет и объяснить что происходит.
````

БАЛЛ:

````python
expenses = {"Еда": 500, "Транспорт": 100}

print(sum(expenses.values()))
print(min(expenses.values()))
print(max(expenses.values()))
````

````python
print(max(expenses))  # Максимум по ключам или по значениям? А вывод ключа или значения?
````

БАЛЛ:

````python
print(max(expenses, key=lambda category: expenses[category])) # Что выведет и почему?
````

БАЛЛ:

````python
scores = {"Ралина": 95, "Ярослав": 80, "Камилла": 95}
best_score = max(scores.values())

for name, score in scores.items():
    if score == best_score:
        print(name)
````

````python
for category in sorted(expenses, key=lambda category: expenses[category], reverse=True): # Объяснить что происходит.
    print(category, expenses[category])
````

БАЛЛ:

````python
expenses = {"Еда": 500, "Транспорт": 100}
total = sum(expenses.values())
shares = {}
large_expenses = []

for category, amount in expenses.items():
    shares[category] = round(amount / total * 100, 1)
    if amount > 200:
        large_expenses.append(category)

# Объяснить что будет выведено. Точно считать не нужно.
print(shares)
print(large_expenses)
````

````python
expenses = {"Еда": 500, "Транспорт": 100}
total = sum(expenses.values())

shares = {category: round(amount / total * 100, 1) for category, amount in expenses.items()}
large_expenses = [category for category, amount in expenses.items() if amount > 200]

print(shares)
print(large_expenses)
````

### Среднеквадратичное отклонение

````python
expenses = {"Еда": 500, "Транспорт": 100, "Кафе": 300}
average = sum(expenses.values()) / len(expenses)
squared_deviations = 0

for amount in expenses.values():
    squared_deviations += (amount - average) ** 2

deviation = (squared_deviations / len(expenses)) ** 0.5
print(round(deviation, 2))
````
