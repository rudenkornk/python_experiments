# Множества

Магазин записывает имя покупателя при каждой покупке.

```python
january = ["Эрик", "Егор", "Эрик", "Анна", "Егор", "Диана"]
february = ["Егор", "Никита", "Егор", "Диана", "Татьяна", "Никита"]

print("Покупок в январе:", len(january)) # Сколько покупок совершили в январе?
print("Первый покупатель:", january[0]) # Кто совершил первую покупку?
print("Анна покупала в январе:", "Анна" in january) # Покупала ли Анна в январе?

example = january.copy()
example.append("Элеонора")
print("После новой покупки:", example) # Как добавить покупку Элеоноры?
```

---

**Задача:** получить список покупателей января без повторов и посчитать их.

```python
january_unique = []

for buyer in january:
    if buyer not in january_unique:
        january_unique.append(buyer)

print("Покупатели января:", january_unique)
print("Разных покупателей:", len(january_unique))
```

---

Отдел маркетинга задаёт вопрос:  
Кто покупал и в январе, и в феврале?

```python
both_list = []
february_only_list = []

for buyer in february:
    if buyer in january and buyer not in both_list:
        both_list.append(buyer)

print("Оба месяца:", both_list)
```

---

### Что приходится делать самим?

- Перебирать покупки циклом.
- Проверять, встречался ли клиент в другом месяце.
- Следить, чтобы в ответе не было повторов.

При проверке `in` список просматривает элементы от самого начала до совпадения или до конца.

**Вопрос:** что изменится, если покупок будут миллионы?
Сгенерируем миллион покупок за месяц, выбирая людей из списка.

**Как думаете, сколько времени займёт сравнение месяцев списками?**

```python
import random
from time import perf_counter

names = [
    "Александр", "Михаил", "Андрей", "Дмитрий", "Сергей",
    "Алексей", "Иван", "Николай", "Павел", "Виктор",
    "Роман", "Олег", "Юрий", "Владимир", "Борис",
    "Мария", "Ольга", "Ирина", "Наталья", "Елена",
    "Светлана", "Тамара", "Людмила", "Нина", "Галина",
    "Валентина", "Лариса", "Вера", "Надежда", "Любовь",
]

families = [
    "Иваненко", "Петренко", "Коваленко", "Савченко", "Бондаренко",
    "Шевченко", "Ткаченко", "Кравченко", "Марченко", "Павленко",
    "Романенко", "Захарченко", "Литвиненко", "Сидоренко", "Остапенко",
    "Василенко", "Дорошенко", "Демченко", "Полищенко", "Федоренко",
    "Гриценко", "Яценко", "Кириленко", "Назаренко", "Мироненко",
    "Пономаренко", "Зинченко", "Емельяненко", "Даниленко", "Андриенко",
    "Макаренко", "Левченко", "Семененко", "Тимошенко", "Харченко",
    "Сергиенко", "Клименко", "Мартыненко", "Стеценко", "Терещенко",
]

people = [
    name + " " + family
    for name in names
    for family in families
]

print("Всего разных покупателей:", len(people))
```

```python
random.seed(42)
benchmark_january = random.choices(people, k=1_000_000)
benchmark_february = random.choices(people, k=1_000_000)

print("Первые 20 покупателей февраля:")
for person in benchmark_february[:20]:
    print(person)
```

```python
start = perf_counter()

benchmark_both = []

for buyer in benchmark_february:
    if buyer in benchmark_january and buyer not in benchmark_both:
        benchmark_both.append(buyer)

elapsed = perf_counter() - start
print(f"Время: {elapsed:.3f} с")
print("Покупали в обоих месяцах:", len(benchmark_both))
print("Первые 10 из них:")
for person in benchmark_both[:10]:
    print(person)
```

**Множество (`set`) — коллекция уникальных элементов.**  
Оно само заботится о том, чтобы повторов не было.  

```python
buyers = {"Анна", "Егор", "Диана"}

print(len(buyers))          # Количество элементов.
print("Анна" in buyers)     # Проверка принадлежности.

buyers.add("Никита")        # Добавление элемента.
buyers.remove("Егор")       # Удаление элемента; ошибка, если его нет.
buyers.discard("Татьяна")   # Удаление без ошибки, если элемента нет.

for buyer in buyers:
    print(buyer)
```

```python
print(set(["Анна", "Егор", "Анна"]))  # Из списка.
print(set(("Анна", "Егор", "Анна")))  # Из кортежа.
```

```python
buyers = {"Анна", "Егор", "Анна"}
buyers.add("Анна")
buyers.remove("Анна")
buyers.add("Диана")
buyers.add("Егор")

print(buyers)  # Что окажется в множестве?
```

---

```python
january_buyers = {"Анна", "Егор", "Диана"}
february_buyers = {"Егор", "Диана", "Никита"}

print(january_buyers & february_buyers)  # Пересечение: покупали в обоих месяцах.
print(january_buyers | february_buyers)  # Объединение: покупали хотя бы в одном месяце.
print(january_buyers - february_buyers)  # Разность: покупали в январе, но не в феврале.
print(february_buyers - january_buyers)  # Разность: покупали в феврале, но не в январе.
print(january_buyers ^ february_buyers)  # Симметрическая разность: покупали ровно в одном месяце.
```

```python
# На балл: какие имена окажутся в результате?
january_buyers = {"Анна", "Егор", "Диана"}
february_buyers = {"Егор", "Диана", "Никита"}
march_buyers = {"Анна", "Никита", "Татьяна"}

result = (january_buyers ^ february_buyers) - march_buyers
result = result | (january_buyers & february_buyers)
print(result)
```

```python
start = perf_counter()

january_set = set(benchmark_january)
february_set = set(benchmark_february)
benchmark_both_set = january_set & february_set

elapsed = perf_counter() - start
print(f"Время: {elapsed:.3f} с")
print("Покупали в обоих месяцах:", len(benchmark_both_set))
print("Первые 10 из них:")
for person in list(benchmark_both_set)[:10]:
    print(person)
```

### Что можно и нельзя хранить в множестве?

```python
# Можно: числа, строки, True и False, кортежи из таких элементов.
values = {42, 3.14, "Анна", ("Егор", "Диана")}
print(values)
```

```python
# Нельзя: списки, словари и обычные множества.
# Кортеж со списком внутри тоже не подойдёт: ("Анна", [1, 2]).
# Класть можно только элементы т.н. хешируемых типов.

values = {["Анна", "Егор"]}
```

# Словари

Анна, Егор, Диана и Никита сдали экзамен по квантовой механике.
Нужно хранить их баллы, узнавать результат по имени и исправлять его после апелляции.

**Задача:** узнать балл Дианы и повысить его на 5 после апелляции.

```python
students = ["Анна", "Егор", "Диана", "Никита"]
scores = [72, 85, 91, 64]

index = students.index("Диана")
print(scores[index])

scores[index] += 5
print(scores[index])
```

```python
# Эрик тоже сдал экзамен.
students.append("Эрик")
scores.append(78)

print(students)
print(scores)
```

Приходится поддерживать соответствие позиций в двух списках.
Если изменить один список и забыть про второй, имена и баллы перепутаются.

Два множества тоже не помогут: они не сохранят связь имени с баллом, а одинаковые баллы исчезнут.

---

**Словарь (`dict`) хранит пары «ключ → значение».**
В нашем случае ключ — имя, значение — балл.

```python
scores = {"Анна": 72, "Егор": 85, "Диана": 91, "Никита": 64}

print(scores["Анна"])      # Значение по ключу.
print(len(scores))         # Количество пар.
print("Диана" in scores)    # Проверка наличия ключа.

scores["Эрик"] = 78         # Добавление новой пары.
scores["Анна"] = 80         # Изменение значения.
del scores["Никита"]        # Удаление пары.
print(scores)
```

### Решаем задачу словарём

```python
scores = {"Анна": 72, "Егор": 85}
scores["Анна"] = 90
scores["Диана"] = 85

print(len(scores))
print(scores)
```

### Ограничения и особенности

Ключи уникальны, а значения могут повторяться.
Чтобы хранить баллы двух разных студентов по имени Анна, нужны разные ключи,
например номера студенческих билетов.

```python
# Обращение по отсутствующему ключу вызывает KeyError.
scores["Татьяна"]
```

```python
# get() позволяет задать ответ для отсутствующего ключа.
print(scores.get("Татьяна"))
print(scores.get("Анна"))
```

```python
# Ключи должны быть хешируемыми: список ключом быть не может.
scores[["Анна", "Егор"]] = 90
```

```python
# Значения могут быть списками: например, баллы за несколько экзаменов.
exam_scores = {"Анна": [72, 80], "Егор": [85, 90]}
print(exam_scores["Анна"])
```

```python
# Множество не сохраняет порядок элементов. Что насчет словаря?

scores = {"Егор": 85, "Анна": 72, "Диана": 91}
scores["Егор"] = 95
scores["Эрик"] = 78
for student in scores:
    print(student)
```

```python
# Обращение идёт по ключу: 0 здесь ключ, а не позиция первого элемента.
scores[0]
```
