### Разбиение задачи: подготовить список и напечатать его

````python
def prepare_names(text):
    names = []
    for name in text.split(','):
        names.append(name.strip())
    return sorted(names)

def print_roster(names):
    for number, name in enumerate(names, start=1):
        print(f'{number}. {name}')

students = prepare_names('Станислав, Екатерина, Никита')
print_roster(students)
````

### Функция -- first class object

Диспетчер сортирует пилотов по высоте полёта. Что получает `sorted` в качестве `key`?

````python
flights = [('Елизавета', 7), ('Егор', 12), ('Милана', 4)]

def altitude(flight):
    return flight[1]

print(altitude(flights[0]))
print(sorted(flights, key=altitude))
````

БАЛЛ:

````python
flights = [('Елизавета', 7), ('Егор', 12), ('Милана', 4)]

print(sorted(flights, key=lambda flight: flight[1]))
````
