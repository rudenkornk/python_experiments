### Какие слова чаще всего встречаются в отчёте о летающих тапочках?

````python
with open("flight_story.txt", encoding="utf-8") as file:
    text = file.read()

print(text.split()[:12])
print(text.split(" ")[:12])
````

БАЛЛ:

### «Тапочки!», тапочки и тапочки-скороходы: что считать одним словом?

````python
punctuation = '.,;:!?«»—()"#'
for token in ["«Тапочки!»", "—", "тапочки-скороходы"]:
    print(repr(token.lower().strip(punctuation)))

words = []
for token in text.split():
    word = token.lower().strip(punctuation)
    if word:
        words.append(word)

print(words[:12])
````

БАЛЛ:

### Сколько раз диспетчер написал «тапочки»?

````python
frequencies = {}
for word in words:
    if word not in frequencies:
        frequencies[word] = 0
    frequencies[word] += 1

print(frequencies)
````

БАЛЛ:

### Зачем убирать «и», «на» и «в» из рейтинга?

````python
stopwords = {"и", "а", "на", "в", "к", "с"}
filtered = {}
for word, count in frequencies.items():
    if word not in stopwords:
        filtered[word] = count

print(sorted(frequencies, key=lambda word: frequencies[word], reverse=True)[:8])
print(sorted(filtered, key=lambda word: filtered[word], reverse=True)[:8])
````

БАЛЛ:

### Частоты по убыванию, слова с одинаковой частотой — по алфавиту

````python
ranking = sorted(filtered.items(), key=lambda pair: pair[1], reverse=True)
print(ranking[:8])

ranking = sorted(filtered.items(), key=lambda pair: (-pair[1], pair[0]))
print(ranking[:8])
````

БАЛЛ:

### Сохраним рейтинг: слово;частота

````python
with open("flight_words.csv", "w", encoding="utf-8") as target:
    for word, count in ranking[:8]:
        print(word, count, sep=";", file=target)

with open("flight_words.csv", encoding="utf-8") as source:
    print(source.read())
````

### Как сохранить рейтинг хештегов #тапочки и #полёт из flight_story.txt?

БАЛЛ:

<details>
<summary>Показать код</summary>

```python
tags = {}
with open("flight_story.txt", encoding="utf-8") as source:
    for line in source:
        for word in line.split():
            if word.startswith("#"):
                if word not in tags:
                    tags[word] = 0
                tags[word] += 1

with open("flight_tags.txt", "w", encoding="utf-8") as target:
    for tag in sorted(tags, key=lambda tag: tags[tag], reverse=True):
        print(f"{tag}: {tags[tag]}", file=target)
```

### Средняя температура тапочек: отдельно днём и ночью

````python
temperatures = []
with open("flight_temperatures.csv", encoding="utf-8") as source:
    next(source)
    for line in source:
        fields = line.strip().split(";")
        temperatures.append(list(map(int, fields[1:])))

day_total = 0
night_total = 0
for day, night in temperatures:
    day_total += day
    night_total += night

print(day_total / len(temperatures))
print(night_total / len(temperatures))
````
