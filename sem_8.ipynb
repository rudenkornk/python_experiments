#### Как устроен UTF-8

```text
1 байт:  0xxxxxxx                              (ASCII)
2 байта: 110xxxxx 10xxxxxx
3 байта: 1110xxxx 10xxxxxx 10xxxxxx
4 байта: 11110xxx 10xxxxxx 10xxxxxx 10xxxxxx
```
`x` — биты кода символа Unicode. Например, русская `А` из имени «Алексей» — это `11010000 10010000`.

Емкость UTF-8 -- около 1'000'000 code points.

#### Термины Unicode

| Термин | Code point примера 1 | Пример 1 | Пример 2 | Пример 3 |
|---|---|---|---|---|

| Code point | `U+0061` | a | ◌̂ | ◌̃ |
| Grapheme cluster | `U+0061 U+0323 U+0302 U+0303` | ậ̃ | ȩ̄́ | ӧ̣́ |
